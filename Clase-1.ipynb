# Panorama General 

Actualmene es la clave de la ciencia de datos

Aplicaciones como: 
- Reconocimiento optico
- Filtros de spam

Tienen la capacida de aprender patrones y automatizar desiciones 

Definicion clasica: campo de estudio que da a los ordenadores la capacidad de aprender

## Componentes clave

- Datos de entrenamiento (E)
- Instancias o muestras
- Modelo
- Tarea (T)
- Medida de rendimiento (P)

El machine learning resulta fundamental en problemas donde **no esxisten algoritmos explicitos escalables**

## Tipos de supervision 

- Aprendizaje supervisado 

    Utiliza datosetiquetados para aprender una funcion que mapea entradas y salidas 

- Aprendizaje no supervisado

    Trabaja con datos sin etiquetas, y se duscan estructuras patrones o representaciones latentes 

- Aprendisaje Auto-supervisado

    Genera senales de supervision a partir de los datos

- Aprendisaje Semi-supervisado

    Combina una pequena cantidad de datos etiquetados con grandes volumenes de datos **NO** etiquetados

- Aprendizaje por refuerzo:

    Entrena agentes que aprenden a tomar decisiones mediante interaccion con un entorno y recompensas




# Aprendizaje supervisado

### Tambien lo podemos conocer como aprendizaje estadistico (Note si el conexto es en ciencia o ingieneria )


Tareas mas comunes

- Clasificacion de noticias
- Filtro de Spam
- Indentificacion de particulas en detectores
- Clasificacion de eventos en rayos cosmicos 

## Metricas Tipicas

- Error cuadratico medio
- Raiz del error

NOTA: **Etiqueta** y **Objetivo** suelen usarse como sinonimos, **etiqueta** es mas comun en clasificacion y **objetivo** en regression
